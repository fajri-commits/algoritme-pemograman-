```r
install.packages("gridExtra", repos="http://cran.us.r-project.org")
install.packages("readxl", repos="http://cran.us.r-project.org")
```

```r
#===============================================================================
# Analisis Data Tingkat Pengangguran Terbuka (TPT) Provinsi 2025
#===============================================================================

library(readxl)
library(ggplot2)
library(gridExtra)

# 1. Import Data
path_file <- "Tingkat Pengangguran Terbuka Menurut Provinsi, 2025.xlsx"
if (file.exists(path_file)) {
  data_raw <- read_excel(path_file, skip = 3, col_names = FALSE)
  colnames(data_raw) <- c("Provinsi", "Februari_2025", "Agustus_2025", "Tahunan")

  # Mengambil 39 baris (38 Provinsi + INDONESIA)
  data_tpt <- data_raw[1:39, c("Provinsi", "Agustus_2025")]
  data_tpt$Agustus_2025 <- as.numeric(data_tpt$Agustus_2025)

  # 2. Hitung Ukuran Pemusatan dan Penyebaran
  get_mode <- function(v) {
    uniqv <- unique(v)
    uniqv[which(tabulate(match(v, uniqv)) == max(tabulate(match(v, uniqv))))]
  }

  mean_val   <- mean(data_tpt$Agustus_2025, na.rm = TRUE)
  median_val <- median(data_tpt$Agustus_2025, na.rm = TRUE)
  mode_val   <- get_mode(data_tpt$Agustus_2025)

  range_val  <- max(data_tpt$Agustus_2025, na.rm = TRUE) - min(data_tpt$Agustus_2025, na.rm = TRUE)
  var_val    <- var(data_tpt$Agustus_2025, na.rm = TRUE)
  sd_val     <- sd(data_tpt$Agustus_2025, na.rm = TRUE)

  cat("\n--- RINGKASAN STATISTIK TPT (N = 39) ---\n")
  cat("Mean             :", round(mean_val, 4), "%\n")
  cat("Median           :", round(median_val, 4), "%\n")
  cat("Modus            :", paste(mode_val, collapse = ", "), "%\n")
  cat("Range            :", round(range_val, 4), "%\n")
  cat("Varians          :", round(var_val, 4), "\n")
  cat("Standar Deviasi  :", round(sd_val, 4), "%\n")

  # 3. Visualisasi Data
  p1 <- ggplot(head(data_tpt[order(-data_tpt$Agustus_2025), ], 10),
               aes(x = reorder(Provinsi, Agustus_2025), y = Agustus_2025)) +
    geom_bar(stat = "identity", fill = "red") +
    coord_flip() +
    labs(title = "Top 10 TPT (Termasuk Indonesia)", x = "Wilayah", y = "TPT (%)") +
    theme_minimal()

  p2 <- ggplot(data_tpt, aes(x = Agustus_2025)) +
    geom_histogram(aes(y = ..density..), binwidth = 0.8, fill = "yellow", color = "white", alpha = 0.7) +
    geom_density(color = "black", size = 1) +
    labs(title = "Histogram Sebaran TPT (N = 39)", x = "TPT (%)", y = "Density") +
    theme_minimal()

  p3 <- ggplot(data_tpt, aes(y = Agustus_2025)) +
    geom_boxplot(fill = "purple", alpha = 0.7) +
    labs(title = "Boxplot TPT", y = "TPT (%)") +
    theme_minimal() +
    theme(axis.text.x = element_blank())

  data_tpt$Kategori <- cut(data_tpt$Agustus_2025,
                           breaks = c(0, 3.5, 5.5, 10),
                           labels = c("Rendah (<3.5%)", "Sedang (3.5-5.5%)", "Tinggi (>5.5%)"))
  pie_df <- as.data.frame(table(data_tpt$Kategori))

  p4 <- ggplot(pie_df, aes(x = "", y = Freq, fill = Var1)) +
    geom_bar(stat = "identity", width = 1) +
    coord_polar("y", start = 0) +
    labs(title = "Proporsi Kategori TPT", fill = "Kategori") +
    theme_void()

  grid.arrange(p1, p2, p3, p4, ncol = 2)
} else {
  cat("File 'Tingkat Pengangguran Terbuka Menurut Provinsi, 2025.xlsx' tidak ditemukan. Silakan upload file tersebut ke Google Colab terlebih dahulu.\n")
}
```
