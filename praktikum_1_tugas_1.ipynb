```r
find_median <- function(t){
  j <- sort(t)
  n <- length(j)
  if (n %% 2 != 0) {
    return(j[(n + 1) / 2])
  } else {
    return((j[n / 2] + j[(n / 2) + 1]) / 2)
  }
}
```

```r
T <- c(2, 8, 4, 9, 24, 51, 74)
find_median(T)
```
