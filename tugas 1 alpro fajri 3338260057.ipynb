```r

cek_ganjil_genap <- function(angka) {
  if (angka %% 2 == 0) {
    return("genap")
  } else {
    return("ganjil")
  }
}

#Uji Coba
print(cek_ganjil_genap(9))
print(cek_ganjil_genap(100))
print(cek_ganjil_genap(2039823947))
```
